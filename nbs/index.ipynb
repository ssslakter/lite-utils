## Usage

````python
import time
from lite_utils.core import run_many_chats, Chat, h2str


chat = Chat("gpt-4o-mini", sp ="Answer only in emojis")


async def coro(chat: Chat, item):
    await chat(item)
    return chat.h

t1 = time.time()
# run chats in parallel
# filter function will skip some items before calling coroutine
res = await run_many_chats(coro, ['say hello', 'say goodbye', 'hi!'], chat, filter_fn=lambda x: '!' not in x)
t2 = time.time()
print(f'{(t2-t1):.2f} sec')
````

### Formatted output

````python
print(h2str(res[0]))
````
